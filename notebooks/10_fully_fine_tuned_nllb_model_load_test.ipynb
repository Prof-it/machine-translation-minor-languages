## Loading and Testing the Fine-Tuned Bidirectional NLLB Model
The notebook successfully demonstrates the deployment and inference testing of a fine-tuned NLLB-200 model designed for bidirectional translation between Odia and German.

### Key Achievements:
* **Model Integration:** Successfully loaded the custom fine-tuned model `abhinandansamal/nllb-200-distilled-600M-finetuned-odia-german-bidirectional` using the Hugging Face `transformers` library.
* **Pipeline Setup:** Established a translation pipeline with `device_map="auto"` to efficiently utilize available hardware (GPU).
* **Bidirectional Translation Verification**

```python
from transformers import pipeline
```

```python
# Load the translation pipeline with your fine-tuned model
model_id = "abhinandansamal/nllb-200-distilled-600M-finetuned-odia-german-bidirectional"
translator = pipeline("translation", model=model_id, device_map="auto")
```

```python
# --- Example 1: Translate Odia to German ---
odia_text = "ଆଜି ପାଗ ବହୁତ ଭଲ ଅଛି।"

german_translation = translator(
    odia_text,
    src_lang="ory_Orya",
    tgt_lang="deu_Latn"
)
print(f"Odia Input: {odia_text}")
print(f"German Output: {german_translation[0]['translation_text']}")
# Expected Output: Heute ist das Wetter sehr gut.
```

```python
# --- Example 2: Translate German to Odia ---
german_text = "Wie ist deine Gesundheit?"

odia_translation = translator(
    german_text,
    src_lang="deu_Latn",
    tgt_lang="ory_Orya"
)
print(f"\nGerman Input: {german_text}")
print(f"Odia Output: {odia_translation[0]['translation_text']}")
# Expected Output: ତୁମର ସ୍ୱାସ୍ଥ୍ୟ ଅବସ୍ଥା କ'ଣ?
```

```python
# --- Example 3: Translate German to Odia ---
german_text = "Die Feuerwehr musste zahlreiche Menschen mit Booten in Sicherheit bringen."

odia_translation = translator(
    german_text,
    src_lang="deu_Latn",
    tgt_lang="ory_Orya"
)
print(f"\nGerman Input: {german_text}")
print(f"Odia Output: {odia_translation[0]['translation_text']}")
```

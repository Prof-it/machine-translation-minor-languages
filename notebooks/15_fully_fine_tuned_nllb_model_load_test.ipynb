## Loading and Testing the Fine-Tuned Bidirectional NLLB Model
The notebook successfully demonstrates the deployment and inference testing of a fine-tuned NLLB-200 model designed for bidirectional translation between Odia and German.

### Key Achievements:
* **Model Integration:** Successfully loaded the custom fine-tuned model `abhinandansamal/nllb-200-distilled-600M-finetuned-odia-german-bidirectional` using the Hugging Face `transformers` library.
* **Pipeline Setup:** Established a translation pipeline with `device_map="auto"` to efficiently utilize available hardware (GPU).
* **Bidirectional Translation Verification**

```python
from transformers import pipeline, AutoTokenizer, AutoModelForSeq2SeqLM
import torch
```

```python
# 1. Configuration
model_id = "abhinandansamal/nllb-200-distilled-600M-full-finetuned-odia-german-bidirectional"

print(f"Loading model from Hugging Face Hub: {model_id}...")
```

```python
# 2. Initialize the pipeline
# We use device_map="auto" to leverage GPU if available
translator = pipeline(
    "translation",
    model=model_id,
    dtype=torch.float16 if torch.cuda.is_available() else torch.float32,
    device_map="auto"
)

# --- Task Prefixes (As used in your training phase) ---
PREFIX_ORI_TO_DEU = "translate Odia to German: "
PREFIX_DEU_TO_ORI = "translate German to Odia: "

# --- Example 1: Translate Odia to German ---
odia_text = "ଆଜି ପାଗ ବହୁତ ଭଲ ଅଛି।"
# We prepend the prefix to match training conditions
input_text_1 = PREFIX_ORI_TO_DEU + odia_text

german_translation = translator(
    input_text_1,
    src_lang="ory_Orya",
    tgt_lang="deu_Latn",
    max_length=512
)

print(f"--- Example 1 ---")
print(f"Odia Input: {odia_text}")
print(f"German Output: {german_translation[0]['translation_text']}")

# --- Example 2: Translate German to Odia (Simple) ---
german_text_2 = "Wie ist deine Gesundheit?"
input_text_2 = PREFIX_DEU_TO_ORI + german_text_2

odia_translation_2 = translator(
    input_text_2,
    src_lang="deu_Latn",
    tgt_lang="ory_Orya",
    max_length=512
)

print(f"\n--- Example 2 ---")
print(f"German Input: {german_text_2}")
print(f"Odia Output: {odia_translation_2[0]['translation_text']}")

# --- Example 3: Translate German to Odia (Complex/Research Sentence) ---
german_text_3 = "Die Feuerwehr musste zahlreiche Menschen mit Booten in Sicherheit bringen."
input_text_3 = PREFIX_DEU_TO_ORI + german_text_3

odia_translation_3 = translator(
    input_text_3,
    src_lang="deu_Latn",
    tgt_lang="ory_Orya",
    max_length=512
)

print(f"\n--- Example 3 ---")
print(f"German Input: {german_text_3}")
print(f"Odia Output: {odia_translation_3[0]['translation_text']}")
```

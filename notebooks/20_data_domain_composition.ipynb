```python
# load JSONL file
df = pd.read_json("/content/drive/MyDrive/Research_Paper_Publication/data/authentic_corpus_final.jsonl", lines=True)
```

```python
# Count occurrences of each domain value exactly as stored
domain_counts = (df["domain"].value_counts().reset_index())
```

```python
# Rename columns
domain_counts.columns = ["Domain", "Count"]
```

```python
# Sort by count (highest first)
domain_counts = domain_counts.sort_values(by="Count", ascending=False).reset_index(drop=True)
```

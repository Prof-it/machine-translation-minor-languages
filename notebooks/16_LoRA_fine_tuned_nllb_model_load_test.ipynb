## Loading and Testing a LoRA Fine-Tuned NLLB Model
This notebook demonstrates the successful loading and inference testing of a parameter-efficient **LoRA (Low-Rank Adaptation)** fine-tuned NLLB-200 model. It highlights the efficiency of the LoRA approach, where only the lightweight adapter weights are loaded on top of the base model.

### Key Achievements:
* **LoRA Adapter Integration:** Successfully loaded the specific LoRA fine-tuned model `abhinandansamal/nllb-200-distilled-600M-LoRA-finetuned-odia-german-bidirectional`.
* **Pipeline Efficiency:** Utilized the standard Hugging Face `pipeline` which automatically handled the merging of the LoRA adapter with the base NLLB model for seamless inference.
* **Bidirectional Translation Verification**

```python
from transformers import pipeline
import torch
```

```python
# 1. Configuration - Use the LoRA Merged Repository
MODEL_ID_LORA = "abhinandansamal/nllb-200-distilled-600M-LoRA-finetuned-odia-german-bidirectional"

# Unified Inference Parameters for Scientific Sync
# Using 512 as the standard generation limit
GEN_CONFIG = {
    "max_length": 512,
    "num_beams": 5,
    "length_penalty": 1.0,
    "early_stopping": True
}
```

```python
# 2. Initialize the pipeline
print(f"🚀 Loading LoRA Merged model: {MODEL_ID_LORA}...")
translator_lora = pipeline(
    "translation",
    model=MODEL_ID_LORA,
    dtype=torch.float16 if torch.cuda.is_available() else torch.float32,
    device_map="auto"
)

# Task Prefixes (Crucial to match your training conditions)
PREFIX_ORI_TO_DEU = "translate Odia to German: "
PREFIX_DEU_TO_ORI = "translate German to Odia: "

# --- Example 1: Translate Odia to German ---
odia_text_1 = "ଆଜି ପାଗ ବହୁତ ଭଲ ଅଛି।"
input_1 = PREFIX_ORI_TO_DEU + odia_text_1

res_1 = translator_lora(
    input_1,
    src_lang="ory_Orya",
    tgt_lang="deu_Latn",
    **GEN_CONFIG # Unpacks max_length, num_beams, etc.
)

print(f"\n--- LoRA Example 1 ---")
print(f"Odia Input: {odia_text_1}")
print(f"German Output: {res_1[0]['translation_text']}")

# --- Example 2: Translate German to Odia ---
german_text_2 = "Wie ist deine Gesundheit?"
input_2 = PREFIX_DEU_TO_ORI + german_text_2

res_2 = translator_lora(
    input_2,
    src_lang="deu_Latn",
    tgt_lang="ory_Orya",
    **GEN_CONFIG
)

print(f"\n--- LoRA Example 2 ---")
print(f"German Input: {german_text_2}")
print(f"Odia Output: {res_2[0]['translation_text']}")

# --- Example 3: Research Case (Firefighters) ---
german_text_3 = "Die Feuerwehr musste zahlreiche Menschen mit Booten in Sicherheit bringen."
input_3 = PREFIX_DEU_TO_ORI + german_text_3

res_3 = translator_lora(
    input_3,
    src_lang="deu_Latn",
    tgt_lang="ory_Orya",
    **GEN_CONFIG
)

print(f"\n--- LoRA Example 3 ---")
print(f"German Input: {german_text_3}")
print(f"Odia Output: {res_3[0]['translation_text']}")
```

## Loading and Testing a LoRA Fine-Tuned NLLB Model
This notebook demonstrates the successful loading and inference testing of a parameter-efficient **LoRA (Low-Rank Adaptation)** fine-tuned NLLB-200 model. It highlights the efficiency of the LoRA approach, where only the lightweight adapter weights are loaded on top of the base model.

### Key Achievements:
* **LoRA Adapter Integration:** Successfully loaded the specific LoRA fine-tuned model `abhinandansamal/nllb-200-distilled-600M-LoRA-finetuned-odia-german-bidirectional`. The logs confirm the system downloaded the base model followed by the small ~9.46 MB `adapter_model.safetensors` file.
* **Pipeline Efficiency:** Utilized the standard Hugging Face `pipeline` which automatically handled the merging of the LoRA adapter with the base NLLB model for seamless inference.
* **Bidirectional Translation Verification**

```python
from transformers import pipeline
```

```python
# Load the translation pipeline with your fine-tuned model
model_id = "abhinandansamal/nllb-200-distilled-600M-LoRA-finetuned-odia-german-bidirectional"
translator = pipeline("translation", model=model_id, device_map="auto")
```

```python
# --- Example 1: Translate Odia to German ---
odia_text = "ଆଜି ପାଗ ବହୁତ ଭଲ ଅଛି।"

german_translation = translator(
    odia_text,
    src_lang="ory_Orya",
    tgt_lang="deu_Latn"
)
print(f"Odia Input: {odia_text}")
print(f"German Output: {german_translation[0]['translation_text']}")
# Expected Output: Das Wetter ist heute sehr gut.
```

```python
# --- Example 2: Translate German to Odia ---
german_text = "Wie ist deine Gesundheit?"

odia_translation = translator(
    german_text,
    src_lang="deu_Latn",
    tgt_lang="ory_Orya"
)
print(f"\nGerman Input: {german_text}")
print(f"Odia Output: {odia_translation[0]['translation_text']}")
# Expected Output: ତୁମର ସ୍ବାସ୍ଥ୍ୟ ଅବସ୍ଥା କଣ?
```

```python
# --- Example 3: Translate German to Odia ---
german_text = "Die Feuerwehr musste zahlreiche Menschen mit Booten in Sicherheit bringen."

odia_translation = translator(
    german_text,
    src_lang="deu_Latn",
    tgt_lang="ory_Orya"
)
print(f"\nGerman Input: {german_text}")
print(f"Odia Output: {odia_translation[0]['translation_text']}")
```

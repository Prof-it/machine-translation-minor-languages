# Model Uploader for Hugging Face Hub
## Objective
The sole objective of this notebook is to take a final, trained translation model from its saved directory in Google Drive and publish it to a new repository on the Hugging Face Hub. This is a critical deployment step that makes the model artifacts centrally available and ready for inference by other applications, such as a Gradio web interface.

## Methodology
The script uses the `huggingface_hub` library to programmatically interact with the Hugging Face platform. It automates the following steps:

1. **Authentication:** Securely logs into a Hugging Face account using an access token.
2. **Repository Creation:** Creates a new, public model repository under the specified user account.
3. **File Upload:** Uploads the entire contents of the local model directory—including the model weights, configuration files, and tokenizer files—to the Hub repository.

## Workflow
1. Mounts Google Drive to access the saved model files.
2. Installs the huggingface_hub library.
3. Prompts the user to log in to their Hugging Face account.
4. Creates a new repository on the Hub with a specified name.
5. Uploads all files from the local model path to the Hub repository.

## Input & Output
* **Input:** A folder in Google Drive containing the saved artifacts of a fine-tuned model (e.g., `NLLB_Odia_German_Best_Merged`).
* **Output:** A new, public model repository on the Hugging Face Hub containing all the uploaded model files.

```python
import os
from huggingface_hub import login, HfApi
from google.colab import userdata
```

```python
# 1. Authenticate with Hugging Face
huggingface_token = userdata.get('HF_TOKEN')
login(token=huggingface_token)
```

```python
# --- GLOBAL CONFIGURATION ---
HF_USERNAME = "abhinandansamal"
api = HfApi()
```

## Deploy Full Fine-tuned NLLB Model

```python
FFT_MODEL_HUB_NAME = "nllb-200-distilled-600M-full-finetuned-odia-german-bidirectional"
FFT_LOCAL_PATH = "/content/drive/MyDrive/Research_Paper_Publication/model/nllb-odia-german-translator_model_final_new_v1"

print(f"\n🚀 Deploying Full Fine-Tuned Model...")
api.create_repo(repo_id=f"{HF_USERNAME}/{FFT_MODEL_HUB_NAME}",
                repo_type="model",
                exist_ok=True)

api.upload_folder(
    folder_path=FFT_LOCAL_PATH,
    repo_id=f"{HF_USERNAME}/{FFT_MODEL_HUB_NAME}",
    repo_type="model",
    commit_message="Initial upload of Full Fine-Tuned NLLB-600M Odia-German model"
)
print(f"✅ FFT Model uploaded: https://huggingface.co/{HF_USERNAME}/{FFT_MODEL_HUB_NAME}")
```

## Deploying LoRA Fine-tuned NLLB Model

```python
# =========================================================
# 3. DEPLOY LoRA (MERGED STANDALONE) NLLB MODEL
# =========================================================
LORA_MODEL_HUB_NAME = "nllb-200-distilled-600M-LoRA-finetuned-odia-german-bidirectional"
# We use the Merged version for standard portability
LORA_LOCAL_PATH = "/content/drive/MyDrive/Research_Paper_Publication/model/NLLB_Odia_German_Best_Merged"

print(f"\n🚀 Deploying Merged LoRA Model...")
api.create_repo(repo_id=f"{HF_USERNAME}/{LORA_MODEL_HUB_NAME}", repo_type="model", exist_ok=True)

api.upload_folder(
    folder_path=LORA_LOCAL_PATH,
    repo_id=f"{HF_USERNAME}/{LORA_MODEL_HUB_NAME}",
    repo_type="model",
    commit_message="Initial upload of Merged LoRA NLLB-600M Odia-German model"
)
print(f"✅ LoRA Model uploaded: https://huggingface.co/{HF_USERNAME}/{LORA_MODEL_HUB_NAME}")

print("\n🎉 All research models have been successfully shared on the Hub!")
```

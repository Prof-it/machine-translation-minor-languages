```python
import os
```

```python
# --- Configuration ---
FILE1_PATH = "/content/drive/MyDrive/Thesis/test/data/raw/authentic_german_corpus_poc.txt"  # Path to first Odia/German file
FILE2_PATH = "/content/drive/MyDrive/Thesis/test/data/raw/synthetic_nllb-200-distilled-600M_german_corpus_poc.txt"  # Path to second Odia/German file
OUTPUT_FILE = "/content/drive/MyDrive/Thesis/test/data/transformed/final_synthetic_nllb-200-distilled-600M_german_corpus_poc_1.txt"  # Path for merged Odia/German corpus
```

```python
def merge_odia_files(file1_path, file2_path, output_file):
    """
    Merge two Odia text files into a single corpus, with blank lines between sentences
    but no trailing blank lines at the end.
    """
    # --- Step 1: Preliminary Checks ---
    if not os.path.exists(file1_path):
        print(f"--- ERROR: Input file '{file1_path}' not found. ---")
        return
    if not os.path.exists(file2_path):
        print(f"--- ERROR: Input file '{file2_path}' not found. ---")
        return

    # --- Step 2: Read Sentences from Both Files ---
    print(f"Reading Odia sentences from '{file1_path}' and '{file2_path}'...")
    sentences = []

    # Read first file
    with open(file1_path, 'r', encoding='utf-8') as f1:
        # Strip whitespace and filter out empty lines
        file1_sentences = [line.strip() for line in f1 if line.strip()]
        sentences.extend(file1_sentences)
        print(f"Found {len(file1_sentences)} sentences in '{file1_path}'.")

    # Read second file
    with open(file2_path, 'r', encoding='utf-8') as f2:
        file2_sentences = [line.strip() for line in f2 if line.strip()]
        sentences.extend(file2_sentences)
        print(f"Found {len(file2_sentences)} sentences in '{file2_path}'.")

    if not sentences:
        print("--- ERROR: Both input files are empty. Please add Odia sentences to at least one file. ---")
        return

    print(f"Total {len(sentences)} sentences to merge.")

    # --- Step 3: Write Merged Sentences to Output File ---
    print(f"Merging and saving to '{output_file}'...")
    with open(output_file, 'w', encoding='utf-8') as f_out:
        # Write sentences, adding \n\n between sentences but not after the last one
        for i, sentence in enumerate(sentences):
            if i < len(sentences) - 1:
                f_out.write(sentence + '\n\n')  # Add blank line between sentences
            else:
                f_out.write(sentence)  # No trailing blank line for the last sentence

    print(f"\n✅ Merging complete! Merged Odia corpus saved in '{output_file}'.")

# --- Step 4: Run the Function ---
merge_odia_files(FILE1_PATH, FILE2_PATH, OUTPUT_FILE)
```

## CSV to JSONL Conversion (German News Data)
This notebook executes a crucial data preprocessing step by converting a raw CSV dataset of German news articles into the machine-learning-friendly **JSONL (JSON Lines)** format. This standardization is essential for downstream tasks such as pre-training or fine-tuning language models on German text.

### Key Achievements:
* **Data Ingestion:** Successfully loaded the German News Dataset sourced from [Kaggle](https://www.kaggle.com/datasets/pqbsbk/german-news-dataset).
* **Format Transformation:**
  * Converted the tabular CSV data into a line-delimited JSON format (`.jsonl`).
  * Used `force_ascii=False` to ensure proper encoding of German characters (umlauts like ä, ö, ü, ß), preserving the linguistic integrity of the text.
  * **Pipeline Integration:** Prepared the data for direct compatibility with Hugging Face `Datasets` or other streaming data loaders used in subsequent training steps.
  
### Workflow Context:
* **Input:** `german_news_data.csv` (Raw tabular data)
* **Process:** Pandas DataFrame $\rightarrow$ JSONL Serialization
* **Output:** `german_news.jsonl` (Model-ready format)

````python
import pandas as pd
````

````python
# ==========================================
# CONFIGURATION
# ==========================================
# Output file for our clean german data
input_csv = "/content/drive/MyDrive/Research_Paper_Publication/test/data/new_data/german_news_data.csv"
output_jsonl = "/content/drive/MyDrive/Research_Paper_Publication/test/data/new_data/german_news.jsonl"
````

````python
# Read the CSV file
# We use utf-8 encoding to support German characters
df = pd.read_csv(input_csv, encoding='utf-8')

# Inspect the dataframe (optional, for debugging)
print("Data Preview:")
print(df.head())
````

````python
# Convert to JSONL format
# orient='records' creates a list of objects (or one object per line with lines=True)
# force_ascii=False ensures German characters are written as is, not as unicode escape sequences
df.to_json(output_jsonl, orient='records', lines=True, force_ascii=False)

print(f"\nSuccessfully converted '{input_csv}' to '{output_jsonl}'")
````
